# 1. Import Required Libraries

```python
import os
import requests
from IPython.display import Markdown, display, update_display
from huggingface_hub import login
from google.colab import userdata
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer, BitsAndBytesConfig, pipeline
import torch
```

# 3. Model Configuration and Availability Check

```python
# Constants
LLAMA = "meta-llama/Llama-3.1-8B-Instruct"


# Check model availability
try:
    AutoTokenizer.from_pretrained(LLAMA)

    print("✓ Model is available and accessible.")
    print(f"Model: {LLAMA}")

except Exception as e:
    print("✗ Unable to access the model.")
    print(f"Model: {LLAMA}")
    print(f"\nError: {e}")
```

# 4. Create a Speech-to-Text Pipeline with Whisper

```python
audio_filename = "denver_extract.mp3"

pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device='cuda',
    return_timestamps=True
)

result = pipe(audio_filename)
transcription = result["text"]
```

# 5. Display the Transcription

```python
display(Markdown(f"Transcription: \n\n{transcription}"))
```

#6. Define System and User Messages

```python
system_message = """
You are an AI meeting minutes assistant.

Your task is to transform meeting transcripts into clear, professional, and well-structured meeting minutes in Markdown format.

Follow these rules:
- Base the minutes strictly on the provided transcript.
- Do not invent, assume, or hallucinate information.
- If the date, location, attendees, owners, or any other information is not mentioned, write "Not specified".
- Preserve important decisions, conclusions, concerns, and disagreements.
- Summarize discussions rather than copying the transcript verbatim.
- Clearly distinguish between decisions, takeaways, and action items.
- For each action item, include the responsible owner when explicitly mentioned. If no owner is mentioned, write "Not specified".
- Keep the minutes concise but informative.
- Use clear Markdown headings and bullet points.
- Do not include code blocks.
"""

user_prompt = f"""
Create professional meeting minutes from the following meeting transcript.

Include the following sections:

# Meeting Minutes

## Summary
Include:
- Meeting title/type
- Date
- Location
- Attendees

## Discussion Points
Summarize the main topics discussed and the important points raised for each topic.

## Key Takeaways
List the most important conclusions, decisions, and insights from the meeting.

## Action Items
Create a list of action items. For each item, include:
- Action
- Owner
- Deadline (if mentioned)

Important:
- Use only information explicitly stated in the transcript.
- Do not guess or invent missing information.
- If information is not available, write "Not specified".
- Keep the output in Markdown format.
- Do not use code blocks.

Transcript:
{transcription}
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]
```

#7. Configure 4-Bit Quantization for Memory Efficiency

```python
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)
```

#8. Load the LLM and Generate Meeting Minutes

```python
# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(LLAMA)

# Use EOS token as padding token
tokenizer.pad_token = tokenizer.eos_token

# Convert chat messages into model inputs
inputs = tokenizer.apply_chat_template(
    messages,
    tokenize=True,
    add_generation_prompt=True,
    return_tensors="pt",
    return_dict=True
)

# Move input tensors to GPU
inputs = {key: value.to("cuda") for key, value in inputs.items()}

# Stream generated tokens
streamer = TextStreamer(
    tokenizer,
    skip_prompt=True
)

# Load the quantized model
model = AutoModelForCausalLM.from_pretrained(
    LLAMA,
    device_map="auto",
    quantization_config=quant_config
)

# Generate meeting minutes
outputs = model.generate(
    **inputs,
    max_new_tokens=2000,
    streamer=streamer
)
```

#9. Decode and Display the Generated Meeting Minutes

```python
# Get the number of input tokens
input_length = inputs["input_ids"].shape[-1]

# Get only the newly generated tokens
generated_tokens = outputs[0][input_length:]

# Decode the generated tokens
response = tokenizer.decode(
    generated_tokens,
    skip_special_tokens=True
)

# Display the meeting minutes
display(Markdown(response))
```

#10. Build the Gradio User Interface

```python
import gradio as gr
import torch
from pathlib import Path


# ============================================================
# Meeting Processing Pipeline
# ============================================================

def process_meeting(audio_file, progress=gr.Progress()):
    """Transcribe the meeting and generate professional minutes."""

    if audio_file is None:
        raise gr.Error("Please upload or record a meeting first.")

    try:
        # ----------------------------------------------------
        # Step 1 — Speech Recognition
        # ----------------------------------------------------

        progress(0.10, desc="🎙️ Loading meeting audio...")

        transcription_result = pipe(audio_file)

        transcription = transcription_result["text"].strip()

        if not transcription:
            raise gr.Error("No speech could be detected in the audio.")

        # ----------------------------------------------------
        # Step 2 — Prepare LLM Prompt
        # ----------------------------------------------------

        progress(0.35, desc="🧠 Preparing meeting context...")

        user_prompt = f"""
Create professional meeting minutes from the following meeting transcript.

Include exactly these sections:

# Meeting Minutes

## Summary
Include:
- Meeting title/type
- Date
- Location
- Attendees

## Discussion Points
Summarize the main topics discussed and the important points raised.

## Key Takeaways
List the most important conclusions, decisions, and insights.

## Action Items
For each action item include:
- Action
- Owner
- Deadline

Important:
- Use only information explicitly stated in the transcript.
- Do not guess or invent missing information.
- If information is unavailable, write "Not specified".
- Keep the output concise but informative.
- Return the result in Markdown.
- Do not use code blocks.

Transcript:
{transcription}
"""

        messages = [
            {
                "role": "system",
                "content": system_message
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ]

        # ----------------------------------------------------
        # Step 3 — Tokenization
        # ----------------------------------------------------

        progress(0.50, desc="🔤 Tokenizing meeting transcript...")

        inputs = tokenizer.apply_chat_template(
            messages,
            return_tensors="pt"
        ).to("cuda")

        # ----------------------------------------------------
        # Step 4 — LLM Generation
        # ----------------------------------------------------

        progress(0.65, desc="🤖 Generating meeting minutes...")

        with torch.inference_mode():

            outputs = model.generate(
                inputs,
                max_new_tokens=2000,
                do_sample=False,
                pad_token_id=tokenizer.eos_token_id
            )

        # ----------------------------------------------------
        # Step 5 — Decode
        # ----------------------------------------------------

        progress(0.90, desc="✨ Formatting final meeting minutes...")

        generated_tokens = outputs[0][inputs.shape[-1]:]

        meeting_minutes = tokenizer.decode(
            generated_tokens,
            skip_special_tokens=True
        ).strip()

        progress(1.0, desc="✅ Meeting analysis completed!")

        return transcription, meeting_minutes

    except Exception as e:

        raise gr.Error(f"Something went wrong: {str(e)}")


# ============================================================
# Custom CSS
# ============================================================

custom_css = """

/* =========================
   Main Container
========================= */

.gradio-container {
    max-width: 1200px !important;
    margin: auto !important;
    padding: 0 24px 40px !important;
}


/* =========================
   Hero Section
========================= */

.hero {
    text-align: center;
    padding: 45px 20px 30px;
}

.badge {
    display: inline-block;
    padding: 8px 16px;
    border-radius: 999px;
    font-size: 13px;
    font-weight: 700;
    margin-bottom: 18px;

    background: rgba(99, 102, 241, 0.12);
    border: 1px solid rgba(99, 102, 241, 0.25);
}

.hero-title {
    font-size: 48px;
    font-weight: 900;
    letter-spacing: -2px;
    line-height: 1.1;
    margin: 0;
}

.gradient-text {
    background: linear-gradient(
        90deg,
        #6366f1,
        #8b5cf6,
        #ec4899
    );

    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
}

.hero-description {
    max-width: 700px;
    margin: 18px auto;
    font-size: 17px;
    line-height: 1.7;
    opacity: 0.7;
}


/* =========================
   Pipeline
========================= */

.pipeline {
    display: flex;
    justify-content: center;
    align-items: center;
    gap: 10px;
    flex-wrap: wrap;
    margin: 10px auto 35px;
}

.pipeline-step {
    padding: 10px 16px;
    border-radius: 12px;
    background: rgba(128,128,128,0.08);
    font-size: 13px;
    font-weight: 600;
}

.arrow {
    opacity: 0.4;
}


/* =========================
   Cards
========================= */

.glass-card {
    border-radius: 20px !important;
    padding: 22px !important;
}


/* =========================
   Main Button
========================= */

.generate-button {
    height: 56px !important;
    border-radius: 14px !important;
    font-size: 16px !important;
    font-weight: 800 !important;
}


/* =========================
   Output
========================= */

.transcription-box textarea {
    font-size: 14px !important;
    line-height: 1.7 !important;
}

.minutes-box {
    border-radius: 18px !important;
    padding: 25px !important;
    min-height: 500px;
}


/* =========================
   Feature Cards
========================= */

.feature-card {
    text-align: center;
    padding: 22px 12px;
}

.feature-icon {
    font-size: 30px;
    margin-bottom: 8px;
}

.feature-title {
    font-weight: 800;
    font-size: 14px;
}

.feature-description {
    font-size: 12px;
    opacity: 0.6;
    margin-top: 5px;
}


/* =========================
   Footer
========================= */

.footer {
    text-align: center;
    margin-top: 40px;
    opacity: 0.5;
    font-size: 12px;
}

footer {
    display: none !important;
}

"""


# ============================================================
# Gradio Application
# ============================================================

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="indigo",
        secondary_hue="purple",
        neutral_hue="slate",
    ),
    css=custom_css,
    title="AI Meeting Intelligence"
) as demo:

    # --------------------------------------------------------
    # HERO
    # --------------------------------------------------------

    gr.HTML("""
    <div class="hero">

        <div class="badge">
            ✨ AI-POWERED MEETING INTELLIGENCE
        </div>

        <h1 class="hero-title">
            Turn Conversations Into
            <br>
            <span class="gradient-text">
                Actionable Meeting Minutes
            </span>
        </h1>

        <p class="hero-description">
            Upload or record a meeting and let AI transform
            your conversation into clear, structured and
            professional meeting minutes in seconds.
        </p>

    </div>
    """)


    # --------------------------------------------------------
    # AI PIPELINE
    # --------------------------------------------------------

    gr.HTML("""
    <div class="pipeline">

        <div class="pipeline-step">
            🎙️ Audio
        </div>

        <div class="arrow">→</div>

        <div class="pipeline-step">
            📝 Whisper
        </div>

        <div class="arrow">→</div>

        <div class="pipeline-step">
            🧠 Llama 3.1
        </div>

        <div class="arrow">→</div>

        <div class="pipeline-step">
            📋 Meeting Minutes
        </div>

    </div>
    """)


    # --------------------------------------------------------
    # INPUT SECTION
    # --------------------------------------------------------

    with gr.Row():

        with gr.Column(
            scale=1,
            elem_classes=["glass-card"]
        ):

            gr.Markdown("## 🎙️ Meeting Recording")

            gr.Markdown(
                "Upload an audio file or record your meeting "
                "directly using your microphone."
            )

            audio_input = gr.Audio(
                sources=["upload", "microphone"],
                type="filepath",
                label="Meeting Audio"
            )

            generate_btn = gr.Button(
                "✨ Generate Meeting Minutes",
                variant="primary",
                elem_classes=["generate-button"]
            )

            clear_btn = gr.ClearButton(
                components=[
                    audio_input
                ],
                value="↺ Clear"
            )


        with gr.Column(
            scale=1,
            elem_classes=["glass-card"]
        ):

            gr.Markdown("## ⚡ AI Pipeline")

            gr.Markdown("""
            **1. Speech Recognition**
            Whisper converts your meeting audio into text.

            **2. Context Understanding**
            The transcript is prepared using structured prompts.

            **3. LLM Generation**
            Llama 3.2 analyzes the conversation.

            **4. Structured Output**
            The final result contains summaries,
            decisions and action items.
            """)


    # --------------------------------------------------------
    # RESULTS
    # --------------------------------------------------------

    gr.Markdown("---")

    gr.Markdown("## 📊 Meeting Analysis")


    with gr.Tabs():

        # ----------------------------------------------------
        # TRANSCRIPTION TAB
        # ----------------------------------------------------

        with gr.Tab("📝 Transcription"):

            transcription_output = gr.Textbox(
                label="Whisper Transcription",
                placeholder=(
                    "Your meeting transcription "
                    "will appear here..."
                ),
                lines=15,
                interactive=False,
                elem_classes=["transcription-box"]
            )


        # ----------------------------------------------------
        # MINUTES TAB
        # ----------------------------------------------------

        with gr.Tab("📋 Meeting Minutes"):

            minutes_output = gr.Markdown(
                """
                ### 👋 Ready to analyze your meeting

                Upload or record an audio file above and click:

                **✨ Generate Meeting Minutes**

                The AI will produce:

                - 📌 Meeting summary
                - 💬 Discussion points
                - 💡 Key takeaways
                - ✅ Action items
                - 👤 Responsible owners
                - 📅 Deadlines when available
                """,
                elem_classes=["minutes-box"]
            )


    # --------------------------------------------------------
    # FEATURES
    # --------------------------------------------------------

    gr.Markdown("---")

    with gr.Row():

        gr.HTML("""
        <div class="feature-card">
            <div class="feature-icon">🎙️</div>
            <div class="feature-title">
                Speech Recognition
            </div>
            <div class="feature-description">
                Powered by Whisper
            </div>
        </div>
        """)

        gr.HTML("""
        <div class="feature-card">
            <div class="feature-icon">🧠</div>
            <div class="feature-title">
                LLM Intelligence
            </div>
            <div class="feature-description">
                Llama 3.1 8B Instruct
            </div>
        </div>
        """)

        gr.HTML("""
        <div class="feature-card">
            <div class="feature-icon">⚡</div>
            <div class="feature-title">
                Memory Efficient
            </div>
            <div class="feature-description">
                4-bit NF4 Quantization
            </div>
        </div>
        """)

        gr.HTML("""
        <div class="feature-card">
            <div class="feature-icon">📋</div>
            <div class="feature-title">
                Structured Output
            </div>
            <div class="feature-description">
                Professional Markdown
            </div>
        """)


    # --------------------------------------------------------
    # FOOTER
    # --------------------------------------------------------

    gr.HTML("""
    <div class="footer">
        Built with 🤗 Hugging Face • Whisper • Llama • PyTorch • Gradio
    </div>
    """)


    # --------------------------------------------------------
    # EVENTS
    # --------------------------------------------------------

    generate_btn.click(
        fn=process_meeting,
        inputs=audio_input,
        outputs=[
            transcription_output,
            minutes_output
        ],
        show_progress="full"
    )


    clear_btn.add(
        [audio_input, transcription_output, minutes_output]
    )


# ============================================================
# Launch
# ============================================================

demo.launch(
    share=True,
    show_error=True
)
```
